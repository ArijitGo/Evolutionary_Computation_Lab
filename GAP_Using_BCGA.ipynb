```python
import random
import time
import numpy as np
import matplotlib.pyplot as plt

# ============================
# Data must be defined BEFORE running this code
# ============================
# cost = [...]
# resource = [...]
# capacity = [...]

try:
    cost_np = np.array(cost)
    resource_np = np.array(resource)
    capacity_np = np.array(capacity)
except NameError:
    raise NameError("Please define cost, resource, and capacity matrices before running.")

NUM_AGENTS = cost_np.shape[0]
NUM_JOBS = cost_np.shape[1]

assert resource_np.shape == (NUM_AGENTS, NUM_JOBS)
assert len(capacity_np) == NUM_AGENTS

print(f"Data loaded: {NUM_AGENTS} agents, {NUM_JOBS} jobs.")

# ============================
# GA PARAMETERS
# ============================
POP_SIZE = 200
GENERATIONS = 50
CROSSOVER_RATE = 0.8
ELITE_SIZE = 2

CHROM_LENGTH = NUM_JOBS * 4
MUTATION_RATE = 1 / CHROM_LENGTH


# ============================
# Decode chromosome
# ============================
def decode_chromosome(chromosome):

    assignment = []

    for j in range(NUM_JOBS):

        start = j * 4
        bits = chromosome[start:start+4]

        val = bits[0]*8 + bits[1]*4 + bits[2]*2 + bits[3]

        agent = val % NUM_AGENTS

        assignment.append(agent)

    return assignment


# ============================
# Encode assignment
# ============================
def encode_assignment(assignment):

    chromosome = []

    for agent in assignment:

        bits = [(agent >> 3) & 1,
                (agent >> 2) & 1,
                (agent >> 1) & 1,
                agent & 1]

        chromosome.extend(bits)

    return chromosome


# ============================
# Fitness
# ============================
def fitness(chromosome):

    assignment = decode_chromosome(chromosome)

    total_profit = 0

    for j, agent in enumerate(assignment):
        total_profit += cost_np[agent, j]

    return total_profit


# ============================
# Repair Operator
# ============================
def repair_chromosome(chromosome):

    assignment = decode_chromosome(chromosome)

    resource_used = np.zeros(NUM_AGENTS)

    for j, agent in enumerate(assignment):
        resource_used[agent] += resource_np[agent, j]

    for a in range(NUM_AGENTS):

        while resource_used[a] > capacity_np[a]:

            jobs = [j for j in range(NUM_JOBS) if assignment[j] == a]

            if not jobs:
                break

            j = random.choice(jobs)

            feasible_agents = []

            for b in range(NUM_AGENTS):

                if b != a and resource_used[b] + resource_np[b, j] <= capacity_np[b]:
                    feasible_agents.append(b)

            if feasible_agents:

                new_agent = random.choice(feasible_agents)

                resource_used[a] -= resource_np[a, j]
                resource_used[new_agent] += resource_np[new_agent, j]

                assignment[j] = new_agent

            else:
                break

    return encode_assignment(assignment)


# ============================
# Feasibility Check
# ============================
def check_feasible(chromosome):

    assignment = decode_chromosome(chromosome)

    resource_used = np.zeros(NUM_AGENTS)

    for j, agent in enumerate(assignment):
        resource_used[agent] += resource_np[agent, j]

    return np.all(resource_used <= capacity_np)


# ============================
# Random chromosome
# ============================
def random_chromosome():

    return [random.randint(0,1) for _ in range(CHROM_LENGTH)]


# ============================
# Initialize population
# ============================
def initialize_population():

    population = []

    for _ in range(POP_SIZE):

        chrom = random_chromosome()
        chrom = repair_chromosome(chrom)

        population.append(chrom)

    return population


# ============================
# Tournament Selection
# ============================
def selection(population, fitnesses, k=3):

    indices = random.sample(range(len(population)), k)

    best_idx = indices[0]
    best_fit = fitnesses[best_idx]

    for idx in indices[1:]:

        if fitnesses[idx] > best_fit:
            best_fit = fitnesses[idx]
            best_idx = idx

    return population[best_idx][:]


# ============================
# Crossover
# ============================
def crossover(parent1, parent2):

    if random.random() < CROSSOVER_RATE:

        point = random.randint(1, CHROM_LENGTH-1)

        child1 = parent1[:point] + parent2[point:]
        child2 = parent2[:point] + parent1[point:]

    else:

        child1 = parent1[:]
        child2 = parent2[:]

    return child1, child2


# ============================
# Mutation
# ============================
def mutate(chromosome):

    for i in range(CHROM_LENGTH):

        if random.random() < MUTATION_RATE:

            chromosome[i] = 1 - chromosome[i]

    return chromosome


# ============================
# Genetic Algorithm
# ============================
def genetic_algorithm():

    population = initialize_population()

    best_fitness_per_gen = []

    best_chromosome = None
    best_fitness = -float('inf')

    for gen in range(GENERATIONS):

        fitnesses = [fitness(ind) for ind in population]

        gen_best = max(fitnesses)
        gen_best_idx = fitnesses.index(gen_best)

        if gen_best > best_fitness:

            best_fitness = gen_best
            best_chromosome = population[gen_best_idx][:]

        best_fitness_per_gen.append(gen_best)

        new_population = []

        sorted_indices = sorted(range(len(fitnesses)),
                                key=lambda i: fitnesses[i],
                                reverse=True)

        for i in range(ELITE_SIZE):

            new_population.append(population[sorted_indices[i]][:])

        while len(new_population) < POP_SIZE:

            p1 = selection(population, fitnesses)
            p2 = selection(population, fitnesses)

            c1, c2 = crossover(p1, p2)

            c1 = mutate(c1)
            c2 = mutate(c2)

            c1 = repair_chromosome(c1)
            c2 = repair_chromosome(c2)

            new_population.append(c1)

            if len(new_population) < POP_SIZE:
                new_population.append(c2)

        population = new_population

    best_chromosome = repair_chromosome(best_chromosome)

    return best_chromosome, best_fitness, best_fitness_per_gen


# ============================
# Convergence Plot Function (Enhanced)
# ============================
def plot_bcga_convergence(all_histories, feasibles, title='Convergence of BCGA'):
    """
    Plot convergence curves for multiple BCGA runs.
    
    Parameters:
        all_histories : list of lists
            Each inner list contains the best profit per generation for one run.
        feasibles : list of bool
            Indicates whether each run ended with a feasible solution.
        title : str
            Title for the plot.
    """
    plt.figure(figsize=(10, 6))
    generations = np.arange(1, GENERATIONS + 1)
    
    # Plot each run's convergence (only if feasible)
    for i, (hist, feasible) in enumerate(zip(all_histories, feasibles)):
        if feasible:
            hist_plot = np.array(hist, dtype=float)
            # In case any -inf appear (should not happen in repair method)
            hist_plot[hist_plot == -np.inf] = np.nan
            plt.plot(generations, hist_plot, alpha=0.3, linewidth=0.8,
                     label=f'Run {i+1}' if i < 5 else "")
    
    # Compute and plot average convergence (only feasible runs)
    feasible_histories = [h for h, f in zip(all_histories, feasibles) if f]
    if feasible_histories:
        avg_history = np.nanmean(feasible_histories, axis=0)
        plt.plot(generations, avg_history, color='black', linewidth=2.5, label='Average (feasible runs)')
    
    plt.xlabel('Generation')
    plt.ylabel('Best Feasible Profit So Far')
    plt.title(title)
    plt.grid(True, alpha=0.3)
    plt.legend(loc='lower right')
    plt.tight_layout()
    plt.savefig('bcga_convergence.png', dpi=150)
    plt.show()


# ============================
# Run BCGA 10 times
# ============================
if __name__ == "__main__":

    num_runs = 20
    base_seed = 42

    results = []
    all_histories = []

    print("Running BCGA 20 times...\n")

    for run in range(num_runs):

        seed = base_seed + run

        random.seed(seed)
        np.random.seed(seed)

        start_time = time.time()

        best_sol, best_val, history = genetic_algorithm()

        elapsed = time.time() - start_time

        feasible = check_feasible(best_sol)

        results.append((run+1, best_val, elapsed, feasible))

        all_histories.append(history)

        print(f"Run {run+1:2d}: Profit = {best_val:.2f}, Time = {elapsed:.2f} s, Feasible = {feasible}")

    profits = [r[1] for r in results]
    times = [r[2] for r in results]
    feasibles = [r[3] for r in results]

    print("\n=== Summary over 20 runs ===")
    print(f"Average profit   : {np.mean(profits):.2f}")
    print(f"Best profit      : {np.max(profits):.2f}")
    print(f"Worst profit     : {np.min(profits):.2f}")
    print(f"Average time     : {np.mean(times):.2f} s")

    # ============================
    # Enhanced Convergence Plot
    # ============================
    plot_bcga_convergence(all_histories, feasibles, title='Convergence of BCGA (Repair Method)')
```
