```python
import random
import time
import numpy as np
import matplotlib.pyplot as plt

try:
    cost_np = np.array(cost)
    resource_np = np.array(resource)
    capacity_np = np.array(capacity)
except NameError:
    raise NameError("Please define cost, resource, and capacity matrices before running this code.")

NUM_AGENTS = cost_np.shape[0]
NUM_JOBS = cost_np.shape[1]

assert resource_np.shape == (NUM_AGENTS, NUM_JOBS), "Resource matrix shape mismatch"
assert len(capacity_np) == NUM_AGENTS, "Capacity length mismatch"
print(f"Data loaded: {NUM_AGENTS} agents, {NUM_JOBS} jobs.")

# ============================
# GA Parameters (dynamic)
# ============================
POP_SIZE = 200
GENERATIONS = 500
CROSSOVER_RATE = 0.8
ELITE_SIZE = 2
PENALTY_COEF = 10000

CHROM_LENGTH = NUM_JOBS * 4
MUTATION_RATE = 1 / CHROM_LENGTH

# ============================
# Helper functions
# ============================
def decode_chromosome(chromosome):
    assignment = []
    for j in range(NUM_JOBS):
        start = j * 4
        bits = chromosome[start:start+4]
        val = bits[0]*8 + bits[1]*4 + bits[2]*2 + bits[3]
        agent = val % NUM_AGENTS
        assignment.append(agent)
    return assignment

def compute_profit(assignment):
    """Return true profit (no penalty)."""
    return sum(cost_np[agent, j] for j, agent in enumerate(assignment))

def compute_resource_usage(assignment):
    usage = np.zeros(NUM_AGENTS, dtype=int)
    for j, agent in enumerate(assignment):
        usage[agent] += resource_np[agent, j]
    return usage

def is_feasible(assignment):
    usage = compute_resource_usage(assignment)
    return np.all(usage <= capacity_np)

def penalized_fitness(chromosome):
    assignment = decode_chromosome(chromosome)
    profit = compute_profit(assignment)
    usage = compute_resource_usage(assignment)
    excess = np.sum(np.maximum(0, usage - capacity_np))
    return profit - PENALTY_COEF * excess

def random_chromosome():
    return [random.randint(0, 1) for _ in range(CHROM_LENGTH)]

def initialize_population():
    return [random_chromosome() for _ in range(POP_SIZE)]

def selection(population, fitnesses, k=3):
    indices = random.sample(range(len(population)), k)
    best_idx = indices[0]
    best_fit = fitnesses[best_idx]
    for idx in indices[1:]:
        if fitnesses[idx] > best_fit:
            best_fit = fitnesses[idx]
            best_idx = idx
    return population[best_idx][:]

def crossover(parent1, parent2):
    if random.random() < CROSSOVER_RATE:
        point = random.randint(1, CHROM_LENGTH - 1)
        child1 = parent1[:point] + parent2[point:]
        child2 = parent2[:point] + parent1[point:]
    else:
        child1 = parent1[:]
        child2 = parent2[:]
    return child1, child2

def mutate(chromosome):
    for i in range(CHROM_LENGTH):
        if random.random() < MUTATION_RATE:
            chromosome[i] = 1 - chromosome[i]
    return chromosome

def genetic_algorithm():
    population = initialize_population()
    best_profit_per_gen = []      # best feasible profit each generation
    best_chromosome = None
    best_profit = -float('inf')
    
    for gen in range(GENERATIONS):
        # Evaluate penalized fitness for selection
        fitnesses = [penalized_fitness(ind) for ind in population]
        
        # Also compute true profit and feasibility for each individual
        profits = []
        feasibles = []
        for ind in population:
            assign = decode_chromosome(ind)
            profits.append(compute_profit(assign))
            feasibles.append(is_feasible(assign))
        
        # Best feasible profit in this generation
        gen_best_profit = -float('inf')
        for p, f in zip(profits, feasibles):
            if f and p > gen_best_profit:
                gen_best_profit = p
        # If no feasible individual, keep previous best (or -inf)
        best_profit_per_gen.append(gen_best_profit)
        
        # Update global best
        if gen_best_profit > best_profit:
            best_profit = gen_best_profit
            # Find the chromosome that gives this profit
            for ind, p, f in zip(population, profits, feasibles):
                if f and p == best_profit:
                    best_chromosome = ind[:]
                    break
        
        # Selection uses penalized fitness
        new_population = []
        sorted_indices = sorted(range(len(fitnesses)), key=lambda i: fitnesses[i], reverse=True)
        for i in range(ELITE_SIZE):
            new_population.append(population[sorted_indices[i]][:])
        
        while len(new_population) < POP_SIZE:
            p1 = selection(population, fitnesses)
            p2 = selection(population, fitnesses)
            c1, c2 = crossover(p1, p2)
            c1 = mutate(c1)
            c2 = mutate(c2)
            new_population.append(c1)
            if len(new_population) < POP_SIZE:
                new_population.append(c2)
        
        population = new_population
    
    return best_chromosome, best_profit, best_profit_per_gen

# ============================
# Run BCGA 20 times and plot profit convergence with average
# ============================
if __name__ == "__main__":
    num_runs = 20
    base_seed = 42
    results = []
    all_profit_histories = []

    print("Running BCGA 20 times...\n")

    for run in range(num_runs):
        seed = base_seed + run
        random.seed(seed)
        np.random.seed(seed)
        
        start_time = time.time()
        best_sol, best_val, profit_history = genetic_algorithm()
        elapsed = time.time() - start_time
        
        # Verify final solution feasibility
        assign = decode_chromosome(best_sol)
        feasible = is_feasible(assign)
        
        results.append((run+1, best_val, elapsed, feasible))
        all_profit_histories.append(profit_history)
        
        print(f"Run {run+1:2d}: Profit = {best_val:.2f}, Time = {elapsed:.2f} s, Feasible = {feasible}")

    # Summary
    profits = [r[1] for r in results]
    times = [r[2] for r in results]
    
    avg_profit = np.mean(profits)
    std_profit = np.std(profits)
    best_profit = np.max(profits)
    worst_profit = np.min(profits)
    avg_time = np.mean(times)
    total_time = np.sum(times)
    
    print("\n=== Summary over 20 runs ===")
    print(f"Average profit   : {avg_profit:.2f} ± {std_profit:.2f}")
    print(f"Best profit      : {best_profit:.2f}")
    print(f"Worst profit     : {worst_profit:.2f}")
    print(f"Average time     : {avg_time:.2f} s")
    print(f"Total time       : {total_time:.2f} s")

    # ============================
    # Profit Convergence Plot with Average
    # ============================
    plt.figure(figsize=(10, 6))
    generations = np.arange(1, GENERATIONS + 1)
    
    # Plot each run's profit history
    for i, history in enumerate(all_profit_histories):
        # Convert -inf to NaN for plotting (if no feasible solution in early gens)
        hist_plot = np.array(history, dtype=float)
        hist_plot[hist_plot == -np.inf] = np.nan
        plt.plot(generations, hist_plot, alpha=0.3, linewidth=0.8,
                 label=f'Run {i+1}' if i < 5 else "")
    
    # Compute average profit across runs (ignoring NaN)
    # Stack histories, then take nanmean along axis 0
    stacked = np.vstack([np.array(h, dtype=float) for h in all_profit_histories])
    avg_history = np.nanmean(stacked, axis=0)
    plt.plot(generations, avg_history, color='black', linewidth=2.5, label='Average Profit')
    
    plt.xlabel('Generation')
    plt.ylabel('Best Profit (Feasible)')
    plt.title('Profit Convergence of BCGA over 20 runs')
    plt.legend(loc='lower right')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```

```python
import random
import time
import numpy as np
import matplotlib.pyplot as plt

try:
    cost_np = np.array(cost)
    resource_np = np.array(resource)
    capacity_np = np.array(capacity)
except NameError:
    raise NameError("Please define cost, resource, and capacity matrices before running this code.")

NUM_AGENTS = cost_np.shape[0]
NUM_JOBS = cost_np.shape[1]

assert resource_np.shape == (NUM_AGENTS, NUM_JOBS), "Resource matrix shape mismatch"
assert len(capacity_np) == NUM_AGENTS, "Capacity length mismatch"
print(f"Data loaded: {NUM_AGENTS} agents, {NUM_JOBS} jobs.")

# ============================
# GA Parameters (dynamic)
# ============================
POP_SIZE = 200
GENERATIONS = 500
CROSSOVER_RATE = 0.8
ELITE_SIZE = 2
PENALTY_COEF = 10000

CHROM_LENGTH = NUM_JOBS * 4
MUTATION_RATE = 1 / CHROM_LENGTH

# ============================
# Helper functions
# ============================
def decode_chromosome(chromosome):
    assignment = []
    for j in range(NUM_JOBS):
        start = j * 4
        bits = chromosome[start:start+4]
        val = bits[0]*8 + bits[1]*4 + bits[2]*2 + bits[3]
        agent = val % NUM_AGENTS
        assignment.append(agent)
    return assignment

def compute_profit(assignment):
    """Return true profit (no penalty)."""
    return sum(cost_np[agent, j] for j, agent in enumerate(assignment))

def compute_resource_usage(assignment):
    usage = np.zeros(NUM_AGENTS, dtype=int)
    for j, agent in enumerate(assignment):
        usage[agent] += resource_np[agent, j]
    return usage

def is_feasible(assignment):
    usage = compute_resource_usage(assignment)
    return np.all(usage <= capacity_np)

def penalized_fitness(chromosome):
    assignment = decode_chromosome(chromosome)
    profit = compute_profit(assignment)
    usage = compute_resource_usage(assignment)
    excess = np.sum(np.maximum(0, usage - capacity_np))
    return profit - PENALTY_COEF * excess

def random_chromosome():
    return [random.randint(0, 1) for _ in range(CHROM_LENGTH)]

def initialize_population():
    return [random_chromosome() for _ in range(POP_SIZE)]

def selection(population, fitnesses, k=3):
    indices = random.sample(range(len(population)), k)
    best_idx = indices[0]
    best_fit = fitnesses[best_idx]
    for idx in indices[1:]:
        if fitnesses[idx] > best_fit:
            best_fit = fitnesses[idx]
            best_idx = idx
    return population[best_idx][:]

def crossover(parent1, parent2):
    if random.random() < CROSSOVER_RATE:
        point = random.randint(1, CHROM_LENGTH - 1)
        child1 = parent1[:point] + parent2[point:]
        child2 = parent2[:point] + parent1[point:]
    else:
        child1 = parent1[:]
        child2 = parent2[:]
    return child1, child2

def mutate(chromosome):
    for i in range(CHROM_LENGTH):
        if random.random() < MUTATION_RATE:
            chromosome[i] = 1 - chromosome[i]
    return chromosome

def genetic_algorithm():
    population = initialize_population()
    best_profit_per_gen = []      # best feasible profit each generation
    best_chromosome = None
    best_profit = -float('inf')
    
    for gen in range(GENERATIONS):
        # Evaluate penalized fitness for selection
        fitnesses = [penalized_fitness(ind) for ind in population]
        
        # Also compute true profit and feasibility for each individual
        profits = []
        feasibles = []
        for ind in population:
            assign = decode_chromosome(ind)
            profits.append(compute_profit(assign))
            feasibles.append(is_feasible(assign))
        
        # === ADDED: Check feasibility counts after first iteration (generation 0) ===
        if gen == 200:
            feasible_count = sum(feasibles)
            infeasible_count = len(feasibles) - feasible_count
            print(f"  After generation 0: Feasible individuals = {feasible_count}, Infeasible = {infeasible_count}")
        # =========================================================================
        
        # Best feasible profit in this generation
        gen_best_profit = -float('inf')
        for p, f in zip(profits, feasibles):
            if f and p > gen_best_profit:
                gen_best_profit = p
        # If no feasible individual, keep previous best (or -inf)
        best_profit_per_gen.append(gen_best_profit)
        
        # Update global best
        if gen_best_profit > best_profit:
            best_profit = gen_best_profit
            # Find the chromosome that gives this profit
            for ind, p, f in zip(population, profits, feasibles):
                if f and p == best_profit:
                    best_chromosome = ind[:]
                    break
        
        # Selection uses penalized fitness
        new_population = []
        sorted_indices = sorted(range(len(fitnesses)), key=lambda i: fitnesses[i], reverse=True)
        for i in range(ELITE_SIZE):
            new_population.append(population[sorted_indices[i]][:])
        
        while len(new_population) < POP_SIZE:
            p1 = selection(population, fitnesses)
            p2 = selection(population, fitnesses)
            c1, c2 = crossover(p1, p2)
            c1 = mutate(c1)
            c2 = mutate(c2)
            new_population.append(c1)
            if len(new_population) < POP_SIZE:
                new_population.append(c2)
        
        population = new_population
    
    return best_chromosome, best_profit, best_profit_per_gen

# ============================
# Run BCGA 20 times and plot profit convergence with average
# ============================
if __name__ == "__main__":
    num_runs = 1
    base_seed = 42
    results = []
    all_profit_histories = []

    print("Running BCGA 20 times...\n")

    for run in range(num_runs):
        seed = base_seed + run
        random.seed(seed)
        np.random.seed(seed)
        
        start_time = time.time()
        best_sol, best_val, profit_history = genetic_algorithm()
        elapsed = time.time() - start_time
        
        # Verify final solution feasibility
        assign = decode_chromosome(best_sol)
        feasible = is_feasible(assign)
        
        results.append((run+1, best_val, elapsed, feasible))
        all_profit_histories.append(profit_history)
        
        print(f"Run {run+1:2d}: Profit = {best_val:.2f}, Time = {elapsed:.2f} s, Feasible = {feasible}\n")

    # Summary
    profits = [r[1] for r in results]
    times = [r[2] for r in results]
    
    avg_profit = np.mean(profits)
    std_profit = np.std(profits)
    best_profit = np.max(profits)
    worst_profit = np.min(profits)
    avg_time = np.mean(times)
    total_time = np.sum(times)
    
    print("\n=== Summary over 20 runs ===")
    print(f"Average profit   : {avg_profit:.2f} ± {std_profit:.2f}")
    print(f"Best profit      : {best_profit:.2f}")
    print(f"Worst profit     : {worst_profit:.2f}")
    print(f"Average time     : {avg_time:.2f} s")
    print(f"Total time       : {total_time:.2f} s")

    # ============================
    # Profit Convergence Plot with Average
    # ============================
    plt.figure(figsize=(10, 6))
    generations = np.arange(1, GENERATIONS + 1)
    
    # Plot each run's profit history
    for i, history in enumerate(all_profit_histories):
        # Convert -inf to NaN for plotting (if no feasible solution in early gens)
        hist_plot = np.array(history, dtype=float)
        hist_plot[hist_plot == -np.inf] = np.nan
        plt.plot(generations, hist_plot, alpha=0.3, linewidth=0.8,
                 label=f'Run {i+1}' if i < 5 else "")
    
    # Compute average profit across runs (ignoring NaN)
    stacked = np.vstack([np.array(h, dtype=float) for h in all_profit_histories])
    avg_history = np.nanmean(stacked, axis=0)
    plt.plot(generations, avg_history, color='black', linewidth=2.5, label='Average Profit')
    
    plt.xlabel('Generation')
    plt.ylabel('Best Profit (Feasible)')
    plt.title('Profit Convergence of BCGA over 20 runs')
    plt.legend(loc='lower right')
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.show()
```
