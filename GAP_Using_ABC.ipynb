```python
import random
import numpy as np
import time
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# The following matrices (cost, resource, capacity) are assumed
# to be defined exactly as in the problem statement.
# They are omitted here as requested.
# ------------------------------------------------------------

NUM_AGENTS = 10
NUM_JOBS = 60
LAMBDA = 1_000_000   # penalty coefficient

def evaluate(solution):
    """Return profit - LAMBDA * total resource violation."""
    profit = 0
    used = [0] * NUM_AGENTS
    for job, agent in enumerate(solution):
        profit += cost[agent][job]
        used[agent] += resource[agent][job]
    violation = sum(max(0, used[i] - capacity[i]) for i in range(NUM_AGENTS))
    return profit - LAMBDA * violation

def true_profit_and_feasibility(solution):
    """Return (profit, feasible_flag) without penalty."""
    profit = 0
    used = [0] * NUM_AGENTS
    for job, agent in enumerate(solution):
        profit += cost[agent][job]
        used[agent] += resource[agent][job]
    feasible = all(used[i] <= capacity[i] for i in range(NUM_AGENTS))
    return profit, feasible

def generate_neighbour(source, partner):
    """Discrete neighbour: copy partner or random agent for one job."""
    new_sol = source.copy()
    job = random.randint(0, NUM_JOBS - 1)
    if random.random() < 0.5:
        new_sol[job] = partner[job]
    else:
        new_sol[job] = random.randint(0, NUM_AGENTS - 1)
    return new_sol

def abc_gap(Np, T, limit, seed=None):
    """Run ABC and return (best_solution, best_true_profit, convergence_history)."""
    if seed is not None:
        random.seed(seed)
        np.random.seed(seed)

    # Initialisation
    population = []
    fitness = []
    trial = []
    for _ in range(Np):
        sol = [random.randint(0, NUM_AGENTS - 1) for _ in range(NUM_JOBS)]
        fit = evaluate(sol)
        population.append(sol)
        fitness.append(fit)
        trial.append(0)

    # Best tracking
    best_idx = np.argmax(fitness)
    best_sol = population[best_idx].copy()
    best_true_profit, _ = true_profit_and_feasibility(best_sol)
    history = []   # best true profit per generation

    for gen in range(T):
        # Employed Bee Phase
        for i in range(Np):
            partner = random.choice([p for p in range(Np) if p != i])
            new_sol = generate_neighbour(population[i], population[partner])
            new_fit = evaluate(new_sol)
            if new_fit > fitness[i]:
                population[i] = new_sol
                fitness[i] = new_fit
                trial[i] = 0
            else:
                trial[i] += 1

        # Probability calculation (shift if necessary)
        total_fit = sum(fitness)
        if total_fit <= 0:
            shift = -total_fit + 1e-6
            prob = [(f + shift) / (total_fit + shift * Np) for f in fitness]
        else:
            prob = [f / total_fit for f in fitness]

        # Onlooker Bee Phase
        m = 0
        n = 0
        while m < Np:
            if random.random() < prob[n]:
                partner = random.choice([p for p in range(Np) if p != n])
                new_sol = generate_neighbour(population[n], population[partner])
                new_fit = evaluate(new_sol)
                if new_fit > fitness[n]:
                    population[n] = new_sol
                    fitness[n] = new_fit
                    trial[n] = 0
                else:
                    trial[n] += 1
                m += 1
            n = (n + 1) % Np

        # Scout Bee Phase
        for i in range(Np):
            if trial[i] > limit:
                new_sol = [random.randint(0, NUM_AGENTS - 1) for _ in range(NUM_JOBS)]
                new_fit = evaluate(new_sol)
                population[i] = new_sol
                fitness[i] = new_fit
                trial[i] = 0

        # Update best solution
        current_best_idx = np.argmax(fitness)
        current_best_sol = population[current_best_idx]
        current_true_profit, _ = true_profit_and_feasibility(current_best_sol)
        if current_true_profit > best_true_profit:
            best_true_profit = current_true_profit
            best_sol = current_best_sol.copy()
        history.append(best_true_profit)

    return best_sol, best_true_profit, history

# ------------------------------------------------------------
# Main: run 20 times, collect statistics, plot convergence
# ------------------------------------------------------------

def main():
    NUM_RUNS = 20
    results = []
    all_histories = []   # store history of best run (or all for averaging)
    best_profit_overall = -float('inf')
    best_solution_overall = None
    best_run_history = None

    print("Running ABC 20 times...\n")
    for run in range(1, NUM_RUNS + 1):
        start_time = time.time()
        sol, profit, history = abc_gap(Np=100, T=1000, limit=100, seed=run)
        elapsed = time.time() - start_time
        _, feasible = true_profit_and_feasibility(sol)
        results.append((profit, elapsed, feasible))
        all_histories.append(history)
        if profit > best_profit_overall:
            best_profit_overall = profit
            best_solution_overall = sol
            best_run_history = history
        print(f"Run {run:2d}: Profit = {profit:.2f}, Time = {elapsed:.2f} s, Feasible = {feasible}")

    # Summary statistics
    profits = [r[0] for r in results]
    times = [r[1] for r in results]
    feasible_count = sum(1 for r in results if r[2])
    avg_profit = np.mean(profits)
    std_profit = np.std(profits)
    best_profit = max(profits)
    worst_profit = min(profits)
    avg_time = np.mean(times)
    total_time = sum(times)

    print("\n=== Summary over {} runs ===".format(NUM_RUNS))
    print(f"Average profit    : {avg_profit:.2f} ± {std_profit:.2f}")
    print(f"Best profit       : {best_profit:.2f}")
    print(f"Worst profit      : {worst_profit:.2f}")
    print(f"Feasible runs     : {feasible_count}/{NUM_RUNS}")
    print(f"Average time      : {avg_time:.2f} s")
    print(f"Total time        : {total_time:.2f} s")

    # Plot convergence graph for the run that achieved the best final profit
    plt.figure(figsize=(10, 5))
    plt.plot(best_run_history, linewidth=2, color='b')
    plt.xlabel('Generation', fontsize=12)
    plt.ylabel('Best True Profit', fontsize=12)
    plt.title(f'Convergence of ABC (Best Run, Final Profit = {best_profit:.0f})', fontsize=14)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig('abc_convergence.png', dpi=150)
    plt.show()

    # Optional: also plot average convergence over all runs
    # (uncomment if desired)
    # max_len = max(len(h) for h in all_histories)
    # padded = [h + [h[-1]]*(max_len - len(h)) for h in all_histories]
    # avg_history = np.mean(padded, axis=0)
    # plt.figure()
    # plt.plot(avg_history, label='Average over 20 runs')
    # plt.xlabel('Generation')
    # plt.ylabel('Average Best Profit')
    # plt.title('Average Convergence')
    # plt.grid()
    # plt.show()

if __name__ == "__main__":
    main()
```

```python
import random
import numpy as np
import time
import matplotlib.pyplot as plt

# ------------------------------------------------------------
# The following matrices (cost, resource, capacity) are assumed
# to be defined exactly as in the problem statement.
# They are omitted here as requested.
# ------------------------------------------------------------

NUM_AGENTS = 10
NUM_JOBS = 60

# ------------------------------------------------------------
# Repair operator: makes an infeasible solution feasible
# ------------------------------------------------------------
def repair(solution):
    """
    Takes an assignment (list of agent indices) and repairs it to satisfy
    all capacity constraints. Returns a new feasible solution (may lose profit).
    Strategy: while any agent is overloaded, pick a random job from that agent
    and reassign it to a random agent that has enough remaining capacity.
    """
    # Work on a copy
    sol = solution.copy()
    # Compute current resource usage
    used = [0] * NUM_AGENTS
    for job, agent in enumerate(sol):
        used[agent] += resource[agent][job]
    
    # Identify overloaded agents
    overloaded = [i for i in range(NUM_AGENTS) if used[i] > capacity[i]]
    
    # Keep trying until all constraints satisfied
    max_iter = 1000
    for _ in range(max_iter):
        if not overloaded:
            break
        # Pick a random overloaded agent
        i = random.choice(overloaded)
        # Find jobs assigned to agent i
        jobs_i = [j for j, a in enumerate(sol) if a == i]
        if not jobs_i:
            overloaded.remove(i)
            continue
        # Pick a random job from that agent
        job = random.choice(jobs_i)
        # Find agents that can take this job without exceeding capacity
        possible_agents = []
        for a in range(NUM_AGENTS):
            if a != i and used[a] + resource[a][job] <= capacity[a]:
                possible_agents.append(a)
        if possible_agents:
            # Move job to a random feasible agent
            new_agent = random.choice(possible_agents)
            used[i] -= resource[i][job]
            used[new_agent] += resource[new_agent][job]
            sol[job] = new_agent
        else:
            # No agent can take this job alone -> try moving a different job
            # Fallback: move to agent with most remaining capacity
            best_agent = None
            best_rem = -1
            for a in range(NUM_AGENTS):
                if a != i:
                    rem = capacity[a] - used[a]
                    if rem > best_rem:
                        best_rem = rem
                        best_agent = a
            if best_agent is not None:
                used[i] -= resource[i][job]
                used[best_agent] += resource[best_agent][job]
                sol[job] = best_agent
        # Recompute overloaded list
        overloaded = [i for i in range(NUM_AGENTS) if used[i] > capacity[i]]
    
    # If still infeasible after random attempts, use a greedy fallback:
    # move all jobs from overloaded agents to the agent with most free capacity.
    if overloaded:
        used = [0] * NUM_AGENTS
        for job, agent in enumerate(sol):
            used[agent] += resource[agent][job]
        overloaded = [i for i in range(NUM_AGENTS) if used[i] > capacity[i]]
        for i in overloaded:
            jobs_i = [j for j, a in enumerate(sol) if a == i]
            # sort jobs by resource consumption descending (move heavy ones first)
            jobs_i.sort(key=lambda j: resource[i][j], reverse=True)
            for job in jobs_i:
                # find agent with most free capacity
                best_agent = None
                best_free = -1
                for a in range(NUM_AGENTS):
                    if a != i:
                        free = capacity[a] - used[a]
                        if free > best_free:
                            best_free = free
                            best_agent = a
                if best_agent is not None:
                    used[i] -= resource[i][job]
                    used[best_agent] += resource[best_agent][job]
                    sol[job] = best_agent
    return sol

def evaluate(solution):
    """Return total profit (assumes solution is feasible)."""
    profit = 0
    for job, agent in enumerate(solution):
        profit += cost[agent][job]
    return profit

def true_profit_and_feasibility(solution):
    """Return (profit, feasible_flag) for a solution."""
    profit = 0
    used = [0] * NUM_AGENTS
    for job, agent in enumerate(solution):
        profit += cost[agent][job]
        used[agent] += resource[agent][job]
    feasible = all(used[i] <= capacity[i] for i in range(NUM_AGENTS))
    return profit, feasible

def generate_neighbour(source, partner):
    """Discrete neighbour: copy partner or random agent for one job, then repair."""
    new_sol = source.copy()
    job = random.randint(0, NUM_JOBS - 1)
    if random.random() < 0.5:
        new_sol[job] = partner[job]
    else:
        new_sol[job] = random.randint(0, NUM_AGENTS - 1)
    # Ensure feasibility
    new_sol = repair(new_sol)
    return new_sol

def abc_gap(Np, T, limit, seed=None):
    """Run ABC with repair-based constraint handling."""
    if seed is not None:
        random.seed(seed)
        np.random.seed(seed)

    # Initialisation: generate random solutions and repair them to be feasible
    population = []
    fitness = []
    trial = []
    for _ in range(Np):
        sol = [random.randint(0, NUM_AGENTS - 1) for _ in range(NUM_JOBS)]
        sol = repair(sol)   # ensure feasibility
        fit = evaluate(sol)
        population.append(sol)
        fitness.append(fit)
        trial.append(0)

    # Best tracking
    best_idx = np.argmax(fitness)
    best_sol = population[best_idx].copy()
    best_true_profit = fitness[best_idx]
    history = []

    for gen in range(T):
        # Employed Bee Phase
        for i in range(Np):
            partner = random.choice([p for p in range(Np) if p != i])
            new_sol = generate_neighbour(population[i], population[partner])
            new_fit = evaluate(new_sol)
            if new_fit > fitness[i]:
                population[i] = new_sol
                fitness[i] = new_fit
                trial[i] = 0
            else:
                trial[i] += 1

        # Probability calculation (fitness values are positive profits)
        total_fit = sum(fitness)
        if total_fit == 0:
            prob = [1.0/Np] * Np
        else:
            prob = [f / total_fit for f in fitness]

        # Onlooker Bee Phase
        m = 0
        n = 0
        while m < Np:
            if random.random() < prob[n]:
                partner = random.choice([p for p in range(Np) if p != n])
                new_sol = generate_neighbour(population[n], population[partner])
                new_fit = evaluate(new_sol)
                if new_fit > fitness[n]:
                    population[n] = new_sol
                    fitness[n] = new_fit
                    trial[n] = 0
                else:
                    trial[n] += 1
                m += 1
            n = (n + 1) % Np

        # Scout Bee Phase
        for i in range(Np):
            if trial[i] > limit:
                # Generate a completely new random solution and repair it
                new_sol = [random.randint(0, NUM_AGENTS - 1) for _ in range(NUM_JOBS)]
                new_sol = repair(new_sol)
                new_fit = evaluate(new_sol)
                population[i] = new_sol
                fitness[i] = new_fit
                trial[i] = 0

        # Update best solution
        current_best_idx = np.argmax(fitness)
        if fitness[current_best_idx] > best_true_profit:
            best_true_profit = fitness[current_best_idx]
            best_sol = population[current_best_idx].copy()
        history.append(best_true_profit)

    return best_sol, best_true_profit, history

# ------------------------------------------------------------
# Main: run 20 times, collect statistics, plot convergence
# ------------------------------------------------------------
def main():
    NUM_RUNS = 20
    results = []
    all_histories = []
    best_profit_overall = -float('inf')
    best_solution_overall = None
    best_run_history = None

    print("Running ABC with repair method 20 times...\n")
    for run in range(1, NUM_RUNS + 1):
        start_time = time.time()
        sol, profit, history = abc_gap(Np=100, T=500, limit=100, seed=run)
        elapsed = time.time() - start_time
        _, feasible = true_profit_and_feasibility(sol)
        results.append((profit, elapsed, feasible))
        all_histories.append(history)
        if profit > best_profit_overall:
            best_profit_overall = profit
            best_solution_overall = sol
            best_run_history = history
        print(f"Run {run:2d}: Profit = {profit:.2f}, Time = {elapsed:.2f} s, Feasible = {feasible}")

    # Summary statistics
    profits = [r[0] for r in results]
    times = [r[1] for r in results]
    feasible_count = sum(1 for r in results if r[2])
    avg_profit = np.mean(profits)
    std_profit = np.std(profits)
    best_profit = max(profits)
    worst_profit = min(profits)
    avg_time = np.mean(times)
    total_time = sum(times)

    print("\n=== Summary over {} runs ===".format(NUM_RUNS))
    print(f"Average profit    : {avg_profit:.2f} ± {std_profit:.2f}")
    print(f"Best profit       : {best_profit:.2f}")
    print(f"Worst profit      : {worst_profit:.2f}")
    print(f"Feasible runs     : {feasible_count}/{NUM_RUNS}")
    print(f"Average time      : {avg_time:.2f} s")
    print(f"Total time        : {total_time:.2f} s")

    # Plot convergence graph for the run that achieved the best final profit
    plt.figure(figsize=(10, 5))
    plt.plot(best_run_history, linewidth=2, color='b')
    plt.xlabel('Generation', fontsize=12)
    plt.ylabel('Best True Profit', fontsize=12)
    plt.title(f'Convergence of ABC with Repair (Best Run, Final Profit = {best_profit:.0f})', fontsize=14)
    plt.grid(True, alpha=0.3)
    plt.tight_layout()
    plt.savefig('abc_repair_convergence.png', dpi=150)
    plt.show()

if __name__ == "__main__":
    main()
```
