```python
import time
import numpy as np
import matplotlib.pyplot as plt

# ======================================================
# Assumptions:
# cost, resource, capacity are already defined as numpy arrays
# with shapes (10,60), (10,60), and (10,) respectively.
# ======================================================

# Problem dimensions
cost = np.array(cost)
resource = np.array(resource)
capacity = np.array(capacity)
m, n = cost.shape  # m = 10, n = 60

# ======================================================
# RCGA parameters
# ======================================================
POP_SIZE = 200
GENERATIONS = 50
ELITE_COUNT = 2

# Bounds for each gene – note: Xu is set slightly below m to avoid floor(m)
LOW, HIGH = 0.0, float(m) - 1e-10

# SBX and Polynomial Mutation parameters
CROSSOVER_RATE = 0.8
ETA_C = 20
MUTATION_RATE = 0.04
ETA_M = 20

# ======================================================
# Global names required by the given formulas
# ======================================================
pop_size = POP_SIZE
n_jobs = n
crossover_rate = CROSSOVER_RATE
eta_c = ETA_C
eta_m = ETA_M
mutation_rate = MUTATION_RATE
Xl = LOW
Xu = HIGH

# ======================================================
# Provided genetic operators (exact code from the user)
# ======================================================
def sbx_crossover(pop):
    new_pop = pop.copy()
    for i in range(0, pop_size, 2):
        if np.random.rand() < crossover_rate:
            p1 = pop[i]
            p2 = pop[i+1]

            u = np.random.rand(n_jobs)

            beta = np.where(
                u <= 0.5,
                (2*u)**(1/(eta_c+1)),
                (1/(2*(1-u)))**(1/(eta_c+1))
            )

            c1 = 0.5*((1+beta)*p1 + (1-beta)*p2)
            c2 = 0.5*((1-beta)*p1 + (1+beta)*p2)

            new_pop[i] = np.clip(c1, Xl, Xu)
            new_pop[i+1] = np.clip(c2, Xl, Xu)
    return new_pop

def polynomial_mutation(pop):
    mutation_mask = np.random.rand(*pop.shape) < mutation_rate
    r = np.random.rand(*pop.shape)

    delta = np.where(
        r < 0.5,
        (2*r)**(1/(eta_m+1)) - 1,
        1 - (2*(1-r))**(1/(eta_m+1))
    )

    pop[mutation_mask] += delta[mutation_mask] * (Xu - Xl)
    return np.clip(pop, Xl, Xu)

# ======================================================
# Helper functions (decoding, evaluation, feasibility)
# ======================================================
def decode(chromosome):
    # Safety: ensure floor values are within [0, m-1]
    return np.minimum(np.floor(chromosome).astype(int), m-1)

def evaluate(chromosome):
    """Return fitness (profit) and profit. Assumes chromosome is feasible."""
    assign = decode(chromosome)
    profit = cost[assign, np.arange(n)].sum()
    return profit, profit  # fitness = profit

def is_feasible(chromosome):
    assign = decode(chromosome)
    used = np.bincount(assign, weights=resource[assign, np.arange(n)], minlength=m)
    return np.all(used <= capacity)

# ======================================================
# Repair heuristic
# ======================================================
def repair(chromosome):
    """
    Greedy repair: move jobs from overloaded agents to agents with spare capacity,
    trying to minimise profit loss. Returns a feasible chromosome.
    If no feasible assignment can be found after many attempts, returns the original
    (which will be detected as infeasible later).
    """
    assign = decode(chromosome)
    used = np.bincount(assign, weights=resource[assign, np.arange(n)], minlength=m)

    # Quick check: if already feasible, return original
    if np.all(used <= capacity):
        return chromosome.copy()

    # Work on copies
    new_chrom = chromosome.copy()
    new_assign = assign.copy()
    new_used = used.copy()

    max_iter = 500  # prevent infinite loops
    for _ in range(max_iter):
        overloaded = np.where(new_used > capacity)[0]
        if len(overloaded) == 0:
            break

        # Pick the most overloaded agent
        i = overloaded[np.argmax(new_used[overloaded] - capacity[overloaded])]
        jobs_i = np.where(new_assign == i)[0]

        best_move = None
        best_profit_change = -np.inf  # we want maximum profit increase

        for j in jobs_i:
            for k in range(m):
                if k == i:
                    continue
                if new_used[k] + resource[k][j] <= capacity[k]:
                    profit_change = cost[k][j] - cost[i][j]  # positive if better
                    if profit_change > best_profit_change:
                        best_profit_change = profit_change
                        best_move = (j, i, k)

        if best_move is None:
            # No feasible move from this agent – cannot repair further
            break

        j, i, k = best_move
        # Perform the move
        new_assign[j] = k
        new_used[i] -= resource[i][j]
        new_used[k] += resource[k][j]
        # Update chromosome so that job j decodes to agent k
        new_chrom[j] = k + 0.5  # any value in [k, k+1) works

    # After loop, if still infeasible, return original (will be penalised later)
    if np.any(new_used > capacity):
        return chromosome.copy()
    return new_chrom

# ======================================================
# Initialization and selection
# ======================================================
def initialize_population():
    pop = np.random.uniform(LOW, HIGH, (POP_SIZE, n))
    # Repair each individual to ensure feasibility from the start
    for i in range(POP_SIZE):
        pop[i] = repair(pop[i])
    return pop

def tournament_selection(pop, fitness, k=3):
    idx = np.random.choice(POP_SIZE, k, replace=False)
    best_idx = idx[np.argmax(fitness[idx])]
    return pop[best_idx].copy()

# ======================================================
# Single GA run (returns best profit, time, feasibility, and convergence history)
# ======================================================
def run_single_ga():
    pop = initialize_population()
    best_profit = -np.inf
    # convergence history: best profit seen so far at each generation
    history = []

    for gen in range(GENERATIONS):
        # Evaluate fitness (profit) for all individuals
        fitness = np.array([evaluate(ind)[0] for ind in pop])

        # Track best solution
        current_best = np.max(fitness)
        if current_best > best_profit:
            best_profit = current_best

        # Record best profit so far
        history.append(best_profit)

        # Elitism: keep the best ELITE_COUNT individuals
        elite_indices = np.argsort(fitness)[-ELITE_COUNT:]
        elites = pop[elite_indices].copy()

        # Create mating pool via tournament selection
        mating_pool = np.array([tournament_selection(pop, fitness) for _ in range(POP_SIZE)])

        # Apply SBX crossover
        offspring = sbx_crossover(mating_pool)

        # Apply polynomial mutation
        offspring = polynomial_mutation(offspring)

        # Repair offspring to ensure feasibility
        for i in range(POP_SIZE):
            offspring[i] = repair(offspring[i])

        # Evaluate offspring fitness
        offspring_fitness = np.array([evaluate(ind)[0] for ind in offspring])

        # Replace the worst ELITE_COUNT offspring with the elites
        worst_idx = np.argsort(offspring_fitness)[:ELITE_COUNT]
        offspring[worst_idx] = elites

        # New population
        pop = offspring

    # After all generations, return best profit (feasible by construction)
    return best_profit, True, history

# ======================================================
# Plot convergence
# ======================================================
def plot_convergence(all_histories, profits, feasibles):
    plt.figure(figsize=(10, 6))
    generations = np.arange(1, GENERATIONS + 1)
    
    # Plot each run's convergence
    for i, hist in enumerate(all_histories):
        plt.plot(generations, hist, alpha=0.3, linewidth=0.8, label=f'Run {i+1}' if i < 5 else "")
    
    # Compute and plot average convergence
    avg_history = np.mean(all_histories, axis=0)
    plt.plot(generations, avg_history, color='black', linewidth=2.5, label='Average')
    
    plt.xlabel('Generation')
    plt.ylabel('Best Profit So Far')
    plt.title('Convergence of Real-Coded GA with Repair')
    plt.grid(True, alpha=0.3)
    plt.legend(loc='lower right')
    plt.tight_layout()
    plt.show()

# ======================================================
# Run 20 times and produce summary
# ======================================================
def main():
    num_runs = 20
    results = []          # (profit, feasible)
    times = []
    all_histories = []    # list of history arrays

    print("Running GA 20 times with repair heuristic...\n")
    for run in range(1, num_runs + 1):
        # Set a different seed for each run to ensure variety
        np.random.seed(run)
        start = time.time()
        profit, feasible, history = run_single_ga()
        elapsed = time.time() - start
        results.append((profit, feasible))
        times.append(elapsed)
        all_histories.append(history)
        print(f"Run {run}: Profit = {profit:.2f}, Time = {elapsed:.2f} s, Feasible = {feasible}")

    # Summary statistics
    profits = [r[0] for r in results]
    feasibles = [r[1] for r in results]
    avg_profit = np.mean(profits)
    std_profit = np.std(profits)
    best_profit = np.max(profits)
    worst_profit = np.min(profits)
    avg_time = np.mean(times)
    total_time = np.sum(times)

    print("\n" + "="*50)
    print("=== Summary over 20 runs ===")
    print(f"Average profit    : {avg_profit:.2f} ± {std_profit:.2f}")
    print(f"Best profit       : {best_profit:.2f}")
    print(f"Worst profit      : {worst_profit:.2f}")
    print(f"Average time      : {avg_time:.2f} s")
    print(f"Total time        : {total_time:.2f} s")

    # Plot convergence
    plot_convergence(all_histories, profits, feasibles)

if __name__ == "__main__":
    main()
```
