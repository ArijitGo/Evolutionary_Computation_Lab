```python
import time
import numpy as np
import matplotlib.pyplot as plt

# Problem dimensions
cost = np.array(cost)
resource = np.array(resource)
capacity = np.array(capacity)
m, n = cost.shape  

# ======================================================
# RCGA parameters
# ======================================================
POP_SIZE = 200
GENERATIONS = 1000
ELITE_COUNT = 2
PENALTY = 1e5

# Bounds for each gene – note: Xu is set slightly below m to avoid floor(m)
LOW, HIGH = 0.0, float(m) - 1e-10

# SBX and Polynomial Mutation parameters
CROSSOVER_RATE = 0.8
ETA_C = 20
MUTATION_RATE = 0.04
ETA_M = 20

# ======================================================
# Global names required by the given formulas
# ======================================================
pop_size = POP_SIZE
n_jobs = n
crossover_rate = CROSSOVER_RATE
eta_c = ETA_C
eta_m = ETA_M
mutation_rate = MUTATION_RATE
Xl = LOW
Xu = HIGH

# ======================================================
# Provided genetic operators (exact code from the user)
# ======================================================
def sbx_crossover(pop):
    new_pop = pop.copy()
    for i in range(0, pop_size, 2):
        if np.random.rand() < crossover_rate:
            p1 = pop[i]
            p2 = pop[i+1]

            u = np.random.rand(n_jobs)

            beta = np.where(
                u <= 0.5,
                (2*u)**(1/(eta_c+1)),
                (1/(2*(1-u)))**(1/(eta_c+1))
            )

            c1 = 0.5*((1+beta)*p1 + (1-beta)*p2)
            c2 = 0.5*((1-beta)*p1 + (1+beta)*p2)

            new_pop[i] = np.clip(c1, Xl, Xu)
            new_pop[i+1] = np.clip(c2, Xl, Xu)
    return new_pop

def polynomial_mutation(pop):
    mutation_mask = np.random.rand(*pop.shape) < mutation_rate
    r = np.random.rand(*pop.shape)

    delta = np.where(
        r < 0.5,
        (2*r)**(1/(eta_m+1)) - 1,
        1 - (2*(1-r))**(1/(eta_m+1))
    )

    pop[mutation_mask] += delta[mutation_mask] * (Xu - Xl)
    return np.clip(pop, Xl, Xu)

# ======================================================
# Helper functions (decoding, evaluation, feasibility)
# ======================================================
def decode(chromosome):
    # Safety: ensure floor values are within [0, m-1]
    return np.minimum(np.floor(chromosome).astype(int), m-1)

def evaluate(chromosome):
    assign = decode(chromosome)
    profit = cost[assign, np.arange(n)].sum()
    used = np.bincount(assign, weights=resource[assign, np.arange(n)], minlength=m)
    violations = np.maximum(0, used - capacity)
    total_violation = violations.sum()
    fitness = profit - PENALTY * total_violation
    return fitness, profit, used, violations

def is_feasible(chromosome):
    assign = decode(chromosome)
    used = np.bincount(assign, weights=resource[assign, np.arange(n)], minlength=m)
    return np.all(used <= capacity)

# ======================================================
# Initialization and selection
# ======================================================
def initialize_population():
    return np.random.uniform(LOW, HIGH, (POP_SIZE, n))

def tournament_selection(pop, fitness, k=3):
    idx = np.random.choice(POP_SIZE, k, replace=False)
    best_idx = idx[np.argmax(fitness[idx])]
    return pop[best_idx].copy()

# ======================================================
# Single GA run (returns best profit, time, feasibility, and convergence history)
# ======================================================
def run_single_ga():
    pop = initialize_population()
    best_feasible = None
    best_feasible_profit = -np.inf
    # convergence history: best feasible profit seen so far at each generation
    history = []

    for gen in range(GENERATIONS):
        # Evaluate fitness
        fitness = np.array([evaluate(ind)[0] for ind in pop])

        # Track best feasible solution
        for i, ind in enumerate(pop):
            if is_feasible(ind):
                prof = evaluate(ind)[1]
                if prof > best_feasible_profit:
                    best_feasible_profit = prof
                    best_feasible = ind.copy()

        # Record current best feasible profit (or -inf if none)
        history.append(best_feasible_profit)

        # Elitism: keep the best ELITE_COUNT individuals
        elite_indices = np.argsort(fitness)[-ELITE_COUNT:]
        elites = pop[elite_indices].copy()

        # Create mating pool via tournament selection
        mating_pool = np.array([tournament_selection(pop, fitness) for _ in range(POP_SIZE)])

        # Apply SBX crossover
        offspring = sbx_crossover(mating_pool)

        # Apply polynomial mutation
        offspring = polynomial_mutation(offspring)

        # Evaluate offspring fitness (for replacement)
        offspring_fitness = np.array([evaluate(ind)[0] for ind in offspring])

        # Replace the worst ELITE_COUNT offspring with the elites
        worst_idx = np.argsort(offspring_fitness)[:ELITE_COUNT]
        offspring[worst_idx] = elites

        # New population
        pop = offspring

    # After all generations
    if best_feasible is not None:
        final_profit = evaluate(best_feasible)[1]
        return final_profit, True, history
    else:
        # No feasible solution found; return best infeasible profit (but mark as infeasible)
        best_idx = np.argmax(fitness)
        best_ind = pop[best_idx]
        profit = evaluate(best_ind)[1]
        return profit, False, history

# ======================================================
# Plot convergence
# ======================================================
def plot_convergence(all_histories, profits, feasibles):
    plt.figure(figsize=(10, 6))
    generations = np.arange(1, GENERATIONS + 1)
    
    # Plot each run's convergence (feasible only, else skip)
    for i, (hist, feasible) in enumerate(zip(all_histories, feasibles)):
        if feasible:
            # Convert -inf to NaN for plotting (these points won't show)
            hist_plot = np.array(hist)
            hist_plot[hist_plot == -np.inf] = np.nan
            plt.plot(generations, hist_plot, alpha=0.3, linewidth=0.8, label=f'Run {i+1}' if i < 5 else "")  # label only first few to avoid legend clutter
    
    # Compute and plot average convergence (only considering feasible runs)
    feasible_histories = [h for h, f in zip(all_histories, feasibles) if f]
    if feasible_histories:
        avg_history = np.nanmean(feasible_histories, axis=0)
        plt.plot(generations, avg_history, color='black', linewidth=2.5, label='Average (feasible runs)')
    
    plt.xlabel('Generation')
    plt.ylabel('Best Feasible Profit So Far')
    plt.title('Convergence of Real-Coded GA')
    plt.grid(True, alpha=0.3)
    plt.legend(loc='lower right')
    plt.tight_layout()
    plt.show()

# ======================================================
# Run 20 times and produce summary
# ======================================================
def main():
    num_runs = 20
    results = []          # (profit, feasible)
    times = []
    all_histories = []    # list of history arrays

    print("Running GA 20 times...\n")
    for run in range(1, num_runs + 1):
        # Set a different seed for each run to ensure variety
        np.random.seed(run)
        start = time.time()
        profit, feasible, history = run_single_ga()
        elapsed = time.time() - start
        results.append((profit, feasible))
        times.append(elapsed)
        all_histories.append(history)
        print(f"Run {run}: Profit = {profit:.2f}, Time = {elapsed:.2f} s, Feasible = {feasible}")

    # Summary statistics
    profits = [r[0] for r in results]
    feasibles = [r[1] for r in results]
    avg_profit = np.mean(profits)
    std_profit = np.std(profits)
    best_profit = np.max(profits)
    worst_profit = np.min(profits)
    avg_time = np.mean(times)
    total_time = np.sum(times)

    print("\n" + "="*50)
    print("=== Summary over 20 runs ===")
    print(f"Average profit    : {avg_profit:.2f} ± {std_profit:.2f}")
    print(f"Best profit       : {best_profit:.2f}")
    print(f"Worst profit      : {worst_profit:.2f}")
    print(f"Average time      : {avg_time:.2f} s")
    print(f"Total time        : {total_time:.2f} s")

    # Plot convergence
    plot_convergence(all_histories, profits, feasibles)

if __name__ == "__main__":
    main()
```
